## Data cleaning 

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df["year"] = df["date"].dt.year
```

```python
df["Month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["dayweek"] = df["date"].dt.day_of_week
df["quarter"] = df["date"].dt.quarter
```

```python
df["quarter"].value_counts()
```

```python
df["department"].value_counts()
```

```python
dep = []
for i in df["department"]:
    dep.append(i.strip())
```

```python
df["department"] = dep
```

```python
df["department"].value_counts()
```

```python
df["department"] = df["department"].map({"sweing":0, "finishing":1})
```

```python
df["wip"].describe()
```

```python
# Fill missing WIP using the median of each specific team
df["wip"] = df.groupby("team")["wip"].transform(lambda x: x.fillna(x.median()))
```

```python
df["class_productive"] = df["actual_productivity"] - df["targeted_productivity"]
```

```python
df["real_class_productive"] = np.where(df["class_productive"]>=0,1,0)
```

```python
df["real_class_productive"].value_counts()
```

```python
df = df.drop(["date","quarter"], axis=1)
df.head()
```

## Regression ML

```python
X = df.drop(["actual_productivity","class_productive","real_class_productive"], axis=1)
y =df["actual_productivity"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test, y_train,y_test = train_test_split(X,y, random_state=42, train_size=0.7)
```

# HYPER MODEL PARAMETER

```python

import warnings
from sklearn.exceptions import ConvergenceWarning

# Suppress FutureWarning (deprecated penalty arguments)
warnings.filterwarnings('ignore', category=FutureWarning)

# Suppress UserWarning (inconsistent penalty/l1_ratio values)
warnings.filterwarnings('ignore', category=UserWarning)

# Suppress ConvergenceWarning (SAG solver max_iter reached)
warnings.filterwarnings('ignore', category=ConvergenceWarning)



```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
import numpy as np
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import GridSearchCV

# 1. Instantiate the regularized linear regression model
elastic_net = ElasticNet(max_iter=5000)

# 2. Define the hyperparameter grid
hyperparameters = {
    # alpha controls the total strength of the penalty (similar to 1/C)
    'alpha': np.logspace(-4, 2, 10),
    
    # l1_ratio controls the blend: 0.0 is pure L2 (Ridge), 1.0 is pure L1 (Lasso)
    'l1_ratio': [0.1, 0.3, 0.5, 0.7, 0.9, 1.0],
    
    'fit_intercept': [True, False]
}

# 3. Set up Grid Search (using Negative MSE as the standard evaluation for regression)
gridsearch = GridSearchCV(
    estimator=elastic_net, 
    param_grid=hyperparameters, 
    cv=5, 
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

# 4. Fit the grid search to find the optimal regression line
gridsearch.fit(X_train, y_train)

# 5. Extract the optimized models
best_params = gridsearch.best_params_
best_model = gridsearch.best_estimator_
```

```python
re_pre_ytest = best_model.predict(X_test) 
```

```python
from sklearn.metrics import r2_score

r2_score(y_test, re_pre_ytest)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV

# Instantiate Vanilla Linear Regression
lr = LinearRegression()

# The only real candidate hyperparameters for basic OLS
hyperparameters = {
    'fit_intercept': [True, False]
}

gridsearch = GridSearchCV(lr, hyperparameters, cv=5, scoring='neg_mean_squared_error')
gridsearch.fit(X_train, y_train)

best_model_linear = gridsearch.best_estimator_
```

```python
linear_ytest = best_model_linear.predict(X_test)
```

```python
class_X = df.drop(["real_class_productive", "class_productive", "actual_productivity"], axis= 1)
class_y = df["real_class_productive"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test, y_train,y_test = train_test_split(class_X,class_y, random_state=42, train_size=0.7, stratify=class_y)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.utils.class_weight import compute_sample_weight
import xgboost as xgb

# 0. Convert your [1, 2, 3] targets into binary 0 or 1
# OPTION A: If you want to group class 1 as '0' and classes 2 & 3 together as '1':
binary_y = np.where(class_y == 1, 0, 1)

# OPTION B: Alternatively, if you want class 3 to be '1' and classes 1 & 2 to be '0':
# binary_y = np.where(class_y == 3, 1, 0)

# 1. Split the data using stratification on the new binary target
X_train, X_test, y_train, y_test = train_test_split(
    class_X, 
    binary_y,           # Using the new binary target array
    random_state=42, 
    train_size=0.7, 
    stratify=binary_y   # Stratifying by binary_y
)

# 2. Calculate sample weights directly using the binary y_train
sample_weights = compute_sample_weight(class_weight='balanced', y=y_train)

# 3. Define the base XGBoost model optimized for binary classification
# Changed objective to binary:logistic and removed num_class
xgb_model = xgb.XGBClassifier(
    objective='binary:logistic', 
    random_state=42, 
    eval_metric='logloss'       # Adjusted to logloss for binary stability
)

# 4. Define your hyperparameter tuning grid
hyperparameters = {
    'max_depth':[1,3,5],
    'learning_rate': [0.01, 0.1, 0.2],
    'n_estimators':[1,3,5],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# 5. Set up Grid Search Cross-Validation
# Using 'f1' instead of 'f1_macro' is standard practice for binary arrays
gridsearch = GridSearchCV(
    estimator=xgb_model,
    param_grid=hyperparameters,
    cv=5,
    scoring='f1',               
    n_jobs=-1,
    verbose=1
)

# 6. Fit the grid search using your sample weights
gridsearch.fit(X_train, y_train, sample_weight=sample_weights)

# 7. Extract the best model configuration
best_params = gridsearch.best_params_
best_model = gridsearch.best_estimator_

print(f"Best parameters found: {best_params}")

# 8. Make predictions (Output will natively be 0 or 1)
y_pred_binary = best_model.predict(X_test)
```

```python
y_pred_binary
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred_binary)
```
